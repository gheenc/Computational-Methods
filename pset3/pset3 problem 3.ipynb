```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: x**3
true_derivative = 27
h = np.logspace(-10, 0, 100)
error = np.abs((f(3 + h) - f(3)) / h - true_derivative)

plt.loglog(h, error)
plt.xlabel("h")
plt.ylabel("|numerical - analytical|")
plt.title("Error in numerical derivative vs h")
plt.show()
```
