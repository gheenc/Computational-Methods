```python
def administer_meds(delta_t, tstop):
    t = 0
    while t < tstop: 
        print(f"Administering meds at t={t}")
        t += delta_t
```

```python
administer_meds(6, 24)
```

```python
administer_meds(2, 10)
```

```python
administer_meds(1, 2)
```

```python
administer_meds(1000, 5000)
```
