```python
 def diagnosis_pt(diagnosis_name):
    code = diagnosis_data[diagnosis_data['long_title']==diagnosis_name]["icd9_code"].item()
    test = icd_data[icd_data["icd9_code"]==code]['subject_id']
    return list(test)
```

```python
patient_data['dob'] = pd.to_datetime(patient_data['dob'])
patient_data['dod'] = pd.to_datetime(patient_data['dod'])
```

```python
def dod_dob(diagnosis_name):
    code = diagnosis_data[diagnosis_data['long_title']==diagnosis_name]['icd9_code'].item()
    test = icd_data[icd_data['icd9_code']==code]['subject_id']
    test2 = list(test)
    dob = patient_data[patient_data['subject_id']==test2]['dob']
    dod = patient_data[patient_data['subject_id']==test2]['dod']
    return(dob-dod)
```

```python
dod_dob('Intestinal infection due to Clostridium difficile')
```
