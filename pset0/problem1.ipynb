```python
#There is ambiguity in that within 1 degree still has 9 data points, 
#ie: 35.1, 35.2, 35.3 etc. 
#so are these normal temperatures or abnormal
```

```python
def temp_tester(normal_temp):
    def actual_temp(temp):
        if (abs(normal_temp - temp) < 1):
            return True
        return False
    return actual_temp
```

```python
human_tester = temp_tester(37)
chicken_tester = temp_tester(41.1)
```
